#### Ejercicio 09.

¿Los precios promedios de "Valdemorillo" y "Galapagar" son iguales? (★★☆)

- Muestra ambos promedios
- Escribe en una celda markdown una conclusión sobre ellos

````python
# Calcular ambos promedios
mean_valdemorillo = df[df['level5'] == "Valdemorillo"]['price'].mean()
mean_galapagar = df[df['level5'] == "Galapagar"]['price'].mean()

# Mostrar ambos promedios
print(f"Precio promedio en Valdemorillo: {mean_valdemorillo:.2f} €")
print(f"Precio promedio en Galapagar: {mean_galapagar:.2f} €")
````

### Conclusión

Al comparar las medias de precio entre ambas poblaciones, se observa que los precios promedios de Valdemorillo y Galapagar no son iguales, existe una diferencia en el valor medio de sus inmuebles de 525.63 €

#### Ejercicio 10

¿Los promedios de precio por metro cuadrado (precio/m2) de "Valdemorillo" y "Galapagar" son iguales? (★★☆)

> Pista: Crea una nueva columna llamada `pps` (*price per square* o precio por metro cuadrado) y luego analiza los valores.

- Muestra ambos promedios de precio por metro cuadrado
- Escribe en una celda markdown una conclusión sobre ellos

````python
# Crear la nueva columna pps
df['pps'] = df['price'] / df['surface']

# Calcular los promedios
pps_valdemorillo = df[df['level5'] == "Valdemorillo"]['pps'].mean()
pps_galapagar = df[df['level5'] == "Galapagar"]['pps'].mean()

# Mostrar ambos promedios
print(f"Precio por m² promedio en Valdemorillo: {pps_valdemorillo:.2f} €/m²")
print(f"Precio por m² promedio en Galapagar: {pps_galapagar:.2f} €/m²")
````

### Conclusión

Al analizar el precio medio por metro cuadrado, se observa que los promedios de Valdemorillo y Galapagar tampoco son iguales, con una diferencia de 288.37 €/m² mas en Galapagar

#### Ejercicio 13

Ahora vamos a trabajar con el "cinturón sur" de Madrid.

Haz un subconjunto del DataFrame original que contenga las siguientes poblaciones (columna level5): "Fuenlabrada", "Leganés", "Getafe", "Alcorcón" (★★☆)

> Pista: Filtra el DataFrame original usando la columna `level5` y la función `isin`.

````python
poblaciones_sur = ["Fuenlabrada", "Leganés", "Getafe", "Alcorcón"]

df_sur = df[df['level5'].isin(poblaciones_sur)]

df_sur.head()
````

#### Ejercicio 14

Calcula la media y la varianza de muestra para las siguientes variables: precio, habitaciones, superficie y baños (★★★)

> Debes usar el subset obtenido en la pregunta 13

- Crea y visualiza un diccionario con todos los valores

````python
# Variables a analizar
columnas = ['price', 'rooms', 'surface', 'bathrooms']

estadisticas_sur = {}

for col in columnas:
    estadisticas_sur[f'{col}_media'] = df_sur[col].mean()
    estadisticas_sur[f'{col}_varianza'] = df_sur[col].var()

estadisticas_sur
````

#### Ejercicio 15

¿Cuál es la casa más cara de cada población del cinturón sur de Madríd? (★★☆)

- Genera un DataFrame con esta información
- Muestra tanto la dirección como el precio de la casa seleccionada de cada población.

#### Ejercicio 16

¿Qué puedes decir acerca del precio por metro cuadrado (precio/m2) entre los municipios de 'Getafe' y 'Alcorcón'? (★★☆)

> Debes usar el subset obtenido en la pregunta 13
>
> Pista: Crea una nueva columna llamada `pps` (price per square en inglés) y luego analiza los valores

````python
df_sur['pps'] = df_sur['price'] / df_sur['surface']

pps_getafe = df_sur[df_sur['level5'] == 'Getafe']['pps'].mean()
pps_alcorcon = df_sur[df_sur['level5'] == 'Alcorcón']['pps'].mean()

print(f"Precio promedio por m² en Getafe: {pps_getafe:.2f} €/m²")
print(f"Precio promedio por m² en Alcorcón: {pps_alcorcon:.2f} €/m²")
````
